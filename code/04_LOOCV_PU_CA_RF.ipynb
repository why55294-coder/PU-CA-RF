```python
import rasterio
import numpy as np
import glob
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# =========================
# 1. 文件路径
# =========================
folder = str(config.FEATURES)

rf_files = sorted(glob.glob(folder + r"\*_RF.tif"))
ca_files = sorted(glob.glob(folder + r"\*_CA.tif"))
all_files = rf_files + ca_files

print("RF:", len(rf_files), "CA:", len(ca_files))

# =========================
# 2. reference grid（统一栅格）
# =========================
ref_path = rf_files[0]

with rasterio.open(ref_path) as ref:
    ref_shape = (ref.height, ref.width)
    profile = ref.profile.copy()
    h, w = ref_shape

# =========================
# 3. 栅格对齐函数（不依赖CRS）
# =========================
def align_no_crs(f):

    with rasterio.open(f) as src:
        data = src.read(1).astype(np.float32)

        out = np.full(ref_shape, np.nan, dtype=np.float32)

        hh = min(h, data.shape[0])
        ww = min(w, data.shape[1])

        out[:hh, :ww] = data[:hh, :ww]

    return out

# =========================
# 4. 构建特征矩阵 X
# =========================
features = []

for f in all_files:
    arr = align_no_crs(f)
    features.append(arr.flatten())

X = np.column_stack(features)

print("X shape:", X.shape)

# =========================
# 5. 读取矿点CSV → labels（全部当矿点）
# =========================
mine_df = pd.read_csv(
    config.SAMPLE_PUL,
    header=None,
    names=["x", "y", "label"]
)


labels = np.zeros(X.shape[0], dtype=int)

with rasterio.open(ref_path) as src:

    width = src.width
    height = src.height

    hit = 0

    for _, row in mine_df.iterrows():

        x = float(row["x"])
        y = float(row["y"])

        try:
            r, c = src.index(x, y)

            if 0 <= r < height and 0 <= c < width:

                idx = r * width + c
                labels[idx] = 1
                hit += 1

            else:
                print("❌越界:", x, y, r, c)

        except Exception as e:
            print("❌失败:", x, y, e)

print("成功写入矿点数:", hit)
print("labels sum:", labels.sum())

# =========================
# 6. mask（去掉NaN像元）
# =========================
mask = ~np.isnan(X).any(axis=1)

X_clean = X[mask]
labels_clean = labels[mask]

P_index = np.where(labels_clean == 1)[0]
U_index = np.where(labels_clean == 0)[0]

print("P:", len(P_index), "U:", len(U_index))
print("labels sum (before mask):", np.sum(labels))
print("labels_clean sum:", np.sum(labels_clean))
print("P_index:", len(P_index))


# =========================
# LOOCV Top20% + Rank验证
# =========================

from sklearn.ensemble import RandomForestClassifier
import numpy as np

T_cv = 100      # 测试建议30，正式可改100

hit_count = 0

rank_list = []

print("\n========== LOOCV开始 ==========")

for i, test_p in enumerate(P_index):

    train_P = P_index[P_index != test_p]

    pred_matrix_cv = np.zeros(
        (X_clean.shape[0], T_cv)
    )

    for t in range(T_cv):

        sampled_U = np.random.choice(
            U_index,
            size=len(train_P),
            replace=False
        )

        train_idx = np.concatenate([
            train_P,
            sampled_U
        ])

        X_train = X_clean[train_idx]

        y_train = np.concatenate([
            np.ones(len(train_P)),
            np.zeros(len(sampled_U))
        ])

        rf_cv = RandomForestClassifier(
            n_estimators=300,
            max_depth=None,
            random_state=t,
            n_jobs=-1
        )

        rf_cv.fit(
            X_train,
            y_train
        )

        pred_matrix_cv[:, t] = (
            rf_cv.predict_proba(X_clean)[:, 1]
        )

    # =========================
    # Bagging平均概率
    # =========================
    prob_map_cv = pred_matrix_cv.mean(axis=1)

    # =========================
    # Top20%命中率
    # =========================
    threshold = np.percentile(
        prob_map_cv,
        80
    )

    test_prob = prob_map_cv[test_p] 

    hit = test_prob >= threshold

    if hit:
        hit_count += 1

    # =========================
    # Rank计算
    # =========================
    rank = (
        np.sum(
            prob_map_cv > test_prob
        )
        / len(prob_map_cv)
    )

    rank_list.append(rank)

    print(
        f"矿点{i+1}: "
        f"概率={test_prob:.4f} "
        f"Rank={rank*100:.2f}% "
        f"{'√命中' if hit else '×未命中'}"
    )

# =========================
# 统计结果
# =========================

rank_list = np.array(rank_list)

hit_rate = (
    hit_count / len(P_index)
)

print("\n========================")
print("LOOCV验证结果")
print("========================")

print(
    f"Top20%命中率: "
    f"{hit_count}/{len(P_index)} "
    f"({hit_rate*100:.2f}%)"
)

print()

print(
    f"平均Rank: "
    f"{rank_list.mean()*100:.2f}%"
)

print(
    f"中位Rank: "
    f"{np.median(rank_list)*100:.2f}%"
)

print(
    f"最佳Rank: "
    f"{rank_list.min()*100:.2f}%"
)

print(
    f"最差Rank: "
    f"{rank_list.max()*100:.2f}%"
)

print("========================\n")


# =========================
# 7. PU Bagging RF
# =========================
T = 100
nP = len(P_index)

pred_matrix = np.zeros((X_clean.shape[0], T))

for t in range(T):

    # 防止样本不足
    sizeU = min(nP, len(U_index))
    sampled_U = np.random.choice(U_index, size=sizeU, replace=False)

    train_idx = np.concatenate([P_index, sampled_U])

    X_train = X_clean[train_idx]
    y_train = np.concatenate([
        np.ones(len(P_index)),
        np.zeros(len(sampled_U))
    ])

    rf = RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        random_state=t,
        n_jobs=-1
    )

    rf.fit(X_train, y_train)

    pred_matrix[:, t] = rf.predict_proba(X_clean)[:, 1]

# =========================
# 8. PU概率 + 不确定性
# =========================
pu_prob = pred_matrix.mean(axis=1)
pu_std = pred_matrix.std(axis=1)

print("PU完成")

# =========================
# 9. 回填栅格
# =========================
full_prob = np.full(X.shape[0], np.nan)
full_std = np.full(X.shape[0], np.nan)

full_prob[mask] = pu_prob
full_std[mask] = pu_std

pu_map = full_prob.reshape(h, w)
std_map = full_std.reshape(h, w)

# =========================
# 9.5 填补空洞
# =========================
valid_mask = ~np.isnan(pu_map)

pu_map[~valid_mask] = np.interp(
    np.flatnonzero(~valid_mask),
    np.flatnonzero(valid_mask),
    pu_map[valid_mask]
)

valid_mask_std = ~np.isnan(std_map)

std_map[~valid_mask_std] = np.interp(
    np.flatnonzero(~valid_mask_std),
    np.flatnonzero(valid_mask_std),
    std_map[valid_mask_std]
)
# =========================
# 10. 输出 GeoTIFF
# =========================
profile.update(
    dtype="float32",
    count=1,
    nodata=-9999,
    compress="lzw"
)

out_dir = config.ensure_dir(config.OUTPUT / "loocv_pu_ca_rf")
out_prob = str(out_dir / "PU_result.tif")
out_std = str(out_dir / "PU_uncertainty.tif")

with rasterio.open(out_prob, "w", **profile) as dst:
    tmp = pu_map.copy()
    tmp[np.isnan(tmp)] = -9999
    dst.write(tmp.astype(np.float32), 1)

with rasterio.open(out_std, "w", **profile) as dst:
    tmp = std_map.copy()
    tmp[np.isnan(tmp)] = -9999
    dst.write(tmp.astype(np.float32), 1)

print("输出完成：PU概率图 + 不确定性图")
```

```python
# =========================
# 11. Prediction-Area (P-A) Curve
# =========================

import matplotlib.pyplot as plt
import pandas as pd

print("=" * 60)
print("开始计算 PU-CA-RF P-A 曲线")
print("=" * 60)

# --------------------------------
# 全区预测概率
# --------------------------------
prob_all = pu_map.flatten()
prob_all = prob_all[~np.isnan(prob_all)]

# --------------------------------
# 矿点预测概率（仅LOOCV保留下来的P_index）
# --------------------------------
mine_prob = pu_prob[P_index]

print("用于验证矿点数量：", len(mine_prob))

# --------------------------------
# 阈值
# --------------------------------
thresholds = np.linspace(
    np.nanmax(prob_all),
    np.nanmin(prob_all),
    200
)

area_percent = []
deposit_percent = []

for th in thresholds:

    # 预测面积百分比
    area = np.sum(prob_all >= th) / len(prob_all) * 100

    # 命中矿点百分比
    deposit = np.sum(mine_prob >= th) / len(mine_prob) * 100

    area_percent.append(area)
    deposit_percent.append(deposit)

area_percent = np.array(area_percent)
deposit_percent = np.array(deposit_percent)

# --------------------------------
# 横坐标递增
# --------------------------------
order = np.argsort(area_percent)

area_percent = area_percent[order]
deposit_percent = deposit_percent[order]

# --------------------------------
# 保存P-A数据
# --------------------------------
pa_df = pd.DataFrame({
    "Predicted Area (%)": area_percent,
    "Deposits Captured (%)": deposit_percent
})

pa_csv = str(config.OUTPUT / "loocv_pu_ca_rf" / "PA_curve.csv")

pa_df.to_csv(pa_csv, index=False)

print("P-A数据已保存：")
print(pa_csv)

# --------------------------------
# 绘图
# --------------------------------
plt.figure(figsize=(6,6))

plt.plot(
    area_percent,
    deposit_percent,
    color="blue",
    linewidth=2,
    label="PU-CA-RF"
)

plt.xlabel(
    "Predicted Area (%)",
    fontsize=13
)

plt.ylabel(
    "Deposits Captured (%)",
    fontsize=13
)

plt.xlim(0,100)
plt.ylim(0,100)

plt.grid(True)

plt.legend(fontsize=11)

plt.tight_layout()

plt.show()

print("P-A曲线完成")
```
